```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import linear_kernel

```

```python
import pandas as pd

# Sample dataset of books
data = {
    "title": [
        "Book A",
        "Book B",
        "Book C",
        "Book D",
        "Book E"
    ],
    "description": [
        "A thrilling adventure in the mountains with unexpected challenges.",
        "A romantic story set in Paris full of emotions and love.",
        "A deep dive into the world of artificial intelligence and machine learning.",
        "An inspiring biography of a famous scientist who changed the world.",
        "A mysterious crime novel involving a detective and hidden clues."
    ]
}

# Create DataFrame
df = pd.DataFrame(data)

# Save to CSV
df.to_csv("books.csv", index=False)

print(" books.csv file created successfully!")
```

```python
# Step 2: Initialize TF-IDF Vectorizer
tfidf = TfidfVectorizer(stop_words='english')

# Step 3: Transform the descriptions into TF-IDF matrix
tfidf_matrix = tfidf.fit_transform(df['description'])

print("TF-IDF Matrix Shape:", tfidf_matrix.shape)

# Step 4: Compute similarity matrix using cosine similarity
cosine_sim = linear_kernel(tfidf_matrix, tfidf_matrix)

# Step 5: Function to recommend top 3 similar books
def recommend_books(title):
    # Check if title exists
    if title not in df['title'].values:
        return f"'{title}' not found in dataset."
    
    idx = df[df['title'] == title].index[0]
    sim_scores = list(enumerate(cosine_sim[idx]))
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)
    
    # Exclude the input book itself (first one)
    sim_scores = sim_scores[1:4]
    
    book_indices = [i[0] for i in sim_scores]
    return df['title'].iloc[book_indices].tolist()

# Example: Recommend books for 'Book A'
print("\nTop 3 Recommendations for 'Book A':")
print(recommend_books("Book A"))
```
