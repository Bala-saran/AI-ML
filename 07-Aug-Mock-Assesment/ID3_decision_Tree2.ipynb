```python
import pandas as pd
import numpy as np
import math
```

```python
data = pd.DataFrame({
    'EmployeeID': ['E1', 'E2', 'E3', 'E4', 'E5'],
    'JobSatisfaction': ['High', 'Low', 'Medium', 'Low', 'High'],
    'WorkLifeBalance': ['Good', 'Poor', 'Good', 'Poor', 'Excellent'],
    'YearsAtCompany': [5, 2, 3, 1, 6],
    'Overtime': ['No', 'Yes', 'No', 'Yes', 'No'],
    'Attrition': ['No', 'Yes', 'No', 'Yes', 'No']
})
```

```python
def entropy(target_col):
    elements, counts = np.unique(target_col, return_counts=True)
    entropy = 0
    for i in range(len(elements)):
        probability = counts[i] / np.sum(counts)
        entropy -= probability * math.log2(probability)
    return round(entropy, 3)
```

```python
def info_gain(data, split_attribute_name, target_name="Attrition"):
    total_entropy = entropy(data[target_name])
    vals, counts = np.unique(data[split_attribute_name], return_counts=True)

    weighted_entropy = 0
    for i in range(len(vals)):
        subset = data[data[split_attribute_name] == vals[i]]
        weighted_entropy += (counts[i] / np.sum(counts)) * entropy(subset[target_name])
        
    gain = total_entropy - weighted_entropy
    return round(gain, 3)
```

```python
print("Entropy of entire dataset (Attrition):", entropy(data['Attrition']))
```

```python
attributes = ['JobSatisfaction', 'WorkLifeBalance', 'Overtime']
```

```python
for attr in attributes:
    print(f"Information Gain for {attr}: {info_gain(data, attr)}")
```

```python
data['YearsBin'] = data['YearsAtCompany'].apply(lambda x: '<=3' if x <= 3 else '>3')
print("Information Gain for YearsAtCompany (binned):", info_gain(data, 'YearsBin'))
```

```python
print("\nDecision Tree (root based on highest gain):")
print("If Overtime == Yes → Attrition = Yes")
print("If Overtime == No  → Attrition = No")
```
